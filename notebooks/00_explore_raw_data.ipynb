```python
import os
from dotenv import load_dotenv
from fredapi import Fred
import pandas as pd
import yfinance as yf

load_dotenv()
fred = Fred(api_key=os.getenv("FRED_API_KEY"))
```

```python
indicator_codes = [
    'T10Y2Y', 'T10Y3M', 'UNRATE', 'ICSA', 'UMCSENT', 'CFNAI', 'GDPC1', 'CPIAUCSL', 'USREC',
    'SAHMREALTIME', 'HOUST', 'PERMIT', 'BAA10Y', 'NFCI', 'PAYEMS', 'INDPRO'
]
for code in indicator_codes:
    try:
        info = fred.get_series_info(code)
        series = fred.get_series(code)
        print(f"--- {code} ---")
        print(info[['title', 'frequency', 'units', 'seasonal_adjustment', 'observation_start', 'observation_end']])
        print(series.head(3))
        print(series.tail(3))
        print(f"dtype: {series.dtype} | nulls: {series.isna().sum()} | rows: {len(series)}")
        print()
    except Exception as e:
        print(f"--- {code}: FAILED — {type(e).__name__}: {e} ---\n")
```
